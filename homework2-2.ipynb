```python
import numpy as np
import scipy.constants as const

# 物理定数
h = const.h        # プランク定数 [J・s]
c = const.c        # 光速 [m/s]
k_B = const.k      # ボルツマン定数 [J/K]

def f(x):
    return 5 * np.exp(-x) + x - 5

# 二分法の初期範囲と目標精度
x1 = 1.0
x2 = 6.0
epsilon = 1e-6

while abs(x2 - x1) > epsilon:
    x_mid = (x1 + x2) / 2.0
    if f(x1) * f(x_mid) > 0:
        x1 = x_mid
    else:
        x2 = x_mid

x_sol = (x1 + x2) / 2.0
b = (h * c) / (k_B * x_sol)

print(f"x = {x_sol:.6f}")
print(f"b = {b:.6e} m K")
```
